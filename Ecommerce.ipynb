```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("ecommerce_sales_analyzer_dataset.csv")
data
```

```python
print(data.isnull().sum())
```

```python
print(data[data["Rating"].isnull()])
```

```python
average_rating = data["Rating"].mean()

print("Average Rating:", average_rating)
```

```python
data["Rating"] = data["Rating"].fillna(average_rating)
data
```

```python
print(data.isnull().sum())
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

```python
print(data.dtypes)
```

```python
data["Order_Date"] = pd.to_datetime(data["Order_Date"])

print(data.dtypes)
```

```python
print("First Order Date:", data["Order_Date"].min())
print("Last Order Date:", data["Order_Date"].max())
```

```python
data["Revenue"] = data["Quantity"] * data["Unit_Price"]

data.head()
```

```python
print(data[["Quantity", "Unit_Price", "Revenue"]])
```

```python
total_revenue = data["Revenue"].sum()

print("Total Revenue:", total_revenue)
```

```python
average_order_value = data["Revenue"].mean()

print("Average Order Value:", average_order_value)
```

```python
highest_order = data.loc[data["Revenue"].idxmax()]

print(highest_order)
```

```python
lowest_order = data.loc[data["Revenue"].idxmin()]

print(lowest_order)
```

```python
product_sales = data.groupby("Product")["Revenue"].sum()

print(product_sales)
```

```python
product_sales = product_sales.sort_values(ascending=False)

print(product_sales)
```

```python
category_sales = data.groupby("Category")["Revenue"].sum()

print(category_sales)
```

```python
category_sales = data.groupby("Category")["Revenue"].sum()

print(category_sales)
```

```python
customer_sales = data.groupby("Customer_Name")["Revenue"].sum()

print(customer_sales)
```

```python
customer_sales = customer_sales.sort_values(ascending=False)

print(customer_sales)
```

```python
city_sales = data.groupby("City")["Revenue"].sum()

print(city_sales)
```

```python
city_sales = data.groupby("City")["Revenue"].sum()

print(city_sales)
```

```python
print(data["Order_Status"].value_counts())
```

```python
print(data["Payment_Method"].value_counts())
```

```python
category_rating = data.groupby("Category")["Rating"].mean()

print(category_rating)
```

```python
data["Month"] = data["Order_Date"].dt.month

print(data[["Order_Date", "Month"]].head())
```

```python
monthly_sales = data.groupby("Month")["Revenue"].sum()

print(monthly_sales)
```

```python
data["Month_Name"] = data["Order_Date"].dt.strftime("%B")

print(data[["Order_Date", "Month_Name"]].head())
```

```python
monthly_sales = data.groupby("Month_Name")["Revenue"].sum()

print(monthly_sales)
```

```python
category_analysis = data.groupby("Category").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Average_Price=("Unit_Price", "mean"),
    Average_Rating=("Rating", "mean")
)

print(category_analysis)
```

```python
category_analysis = category_analysis.sort_values(
    "Total_Revenue",
    ascending=False
)

print(category_analysis)
```

```python
best_product = data.groupby("Product")["Revenue"].sum().idxmax()

print("Best Product:", best_product)
```

```python
best_category = data.groupby("Category")["Revenue"].sum().idxmax()

print("Best Category:", best_category)
```

```python
best_city = data.groupby("City")["Revenue"].sum().idxmax()

print("Best City:", best_city)
```

```python
print("========== E-COMMERCE SALES SUMMARY ==========")

print("Total Revenue:", data["Revenue"].sum())

print("Total Quantity Sold:", data["Quantity"].sum())

print("Average Order Value:", data["Revenue"].mean())

print("Average Rating:", data["Rating"].mean())

print("Best Product:", data.groupby("Product")["Revenue"].sum().idxmax())

print("Best Category:", data.groupby("Category")["Revenue"].sum().idxmax())

print("Best City:", data.groupby("City")["Revenue"].sum().idxmax())

print("Most Used Payment Method:", data["Payment_Method"].value_counts().idxmax())
```
